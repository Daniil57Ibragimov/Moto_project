```python
import pandas as pd
```

```python
data = pd.read_csv('drom.csv', ',')
data
```

```python
del data['Unnamed: 0']
```

```python
data
```

```python
for i in range(len(data)):
    data['Стоимость'][i] = int(data['Стоимость'][i][:-2])
    data['Год выпуска'][i] = int(data['Год выпуска'][i])
    data['Объём двигателся'][i] = int(data['Объём двигателся'][i][:-9])

    if 'км' in data['Пробег'][i]:
        data['Пробег'][i] = int(data['Пробег'][i][:-3])
    elif data['Год выпуска'][i] > 2018:
        data['Пробег'][i] = 0
    
    if data['Тактность'][i] != '-':
        data['Тактность'][i] = int(data['Тактность'][i][:-8])
```
