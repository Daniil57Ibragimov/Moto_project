```python
import pandas as pd

df = pd.read_csv("drom.csv")

df
```

```python
if 'Unnamed: 0' in df.columns:
    df.drop(columns=['Unnamed: 0'], inplace=True)

df
```

```python
df['Стоимость'] = (
    df['Стоимость']
    .str.replace('₽', '', regex=False)
    .str.replace(' ', '', regex=False)
    .astype(int)
)

df
```

```python
df['Год выпуска'] = pd.to_numeric(df['Год выпуска'], errors='coerce').astype('Int64')

df
```

```python
import numpy as np

df['Пробег'] = df['Пробег'].str.replace(' км', '', regex=False).str.replace(' ', '', regex=False)

mask_2020_plus = (df['Год выпуска'].notna()) & (df['Год выпуска'] >= 2020)
mask_2019_or_missing = ~mask_2020_plus

df.loc[mask_2020_plus & (df['Пробег'] == '-'), 'Пробег'] = '0'
df.loc[mask_2019_or_missing & (df['Пробег'] == '-'), 'Пробег'] = np.nan

df['Пробег'] = pd.to_numeric(df['Пробег'], errors='coerce')

df
```

```python
df['Объём двигателя'] = df['Объём двигателя'].astype(str)

df['Объём двигателя'] = (
    df['Объём двигателя']
    .str.replace('куб. см.', '', regex=False)
    .str.replace(' ', '', regex=False)
)

df['Объём двигателя'] = pd.to_numeric(df['Объём двигателя'], errors='coerce').astype('Int64')

df
```

```python
cols_to_clean = ['Тип топлива', 'Тип подачи топлива', 'Коробка передач', 'Тактность']

df[cols_to_clean] = df[cols_to_clean].replace("-", np.nan)

df['Объём двигателя'] = df['Объём двигателя'].replace("-", np.nan)

for col in cols_to_clean:
    df[col] = df[col].astype('category')

df
```

```python
df['Возраст мотоцикла'] = 2025 - df['Год выпуска']

df
```

```python
df.drop(columns=['Год выпуска'], inplace=True)

df
```

```python
df['Коробка передач'] = df['Коробка передач'].fillna('механика')

df
```

```python
df['Коробка (код)'] = df['Коробка передач'].map({
    'механика': 0,
    'автомат': 1
})

df
```

```python
df.drop(columns=['Коробка передач'], inplace=True)

df
```

```python
df['Тактность (код)'] = df['Тактность'].map({
    '2-тактный': 0,
    '4-тактный': 1
})

df
```

```python
df.drop(columns=['Тактность'], inplace=True)

df
```

```python
df['Подача топлива (код)'] = df['Тип подачи топлива'].map({
    'карбюратор': 0,
    'инжектор': 1
})

df
```

```python
df.drop(columns=['Тип подачи топлива'], inplace=True)

df
```

```python
df = df[~df['Тип топлива'].isin(['дизель', 'газ', 'электро'])]

df
```

```python
df['Тип топлива'] = df['Тип топлива'].fillna('бензин')

df
```

```python
import re

df['Модель_clean'] = (
    df['Модель'].astype(str)
    .str.lower()
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True)
)

df_filtered = df[~df['Модель_clean'].str.match(r'^(.+)\s+\1$', na=False)]

df_filtered = df_filtered.drop(columns=['Модель_clean'])

df_filtered = df_filtered.reset_index(drop=True)

df_filtered
```

```python
def extract_brand(model):
    model = str(model).lower()
    if model.startswith("mv agusta"):
        return "MV Agusta"
    elif model.startswith("royal enfield"):
        return "Royal Enfield"
    elif model.startswith("harley-davidson"):
        return "Harley-Davidson"
    elif model.startswith("cfmoto"):
        return "CFMoto"
    else:
        return model.split()[0].capitalize()

df_filtered['Бренд'] = df_filtered['Модель'].apply(extract_brand)

selected_brands = {
    'Ducati', 'MV Agusta', 'Harley-Davidson', 'BMW', 'Indian',
    'Honda', 'Yamaha', 'Suzuki', 'Kawasaki', 'KTM',
    'Royal Enfield', 'CFMoto', 'Triumph', 'Benelli'
}

df_filtered['Бренд'] = df_filtered['Бренд'].apply(lambda b: b if b in selected_brands else "Другое")

df_filtered = pd.get_dummies(df_filtered, columns=['Бренд'], prefix='Бренд', dtype=int)
```
